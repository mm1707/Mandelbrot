```python
import matplotlib.pyplot as plt
import pandas as pd

seq = pd.read_csv("Mandelbrot_sequential.csv")
seq["time_s"] = seq["time_ns"] / 10**9

plt.plot(seq["pixels"], seq["time_s"], marker='o')
plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot sequentially vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.show()
```

```python
thread = pd.read_csv("Mandelbrot_threads.csv")
thread["time_s"] = thread["time_ns"] / 10**9

plt.plot(thread["pixels"], thread["time_s"], marker='o')
plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot with Threads vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.show()
```

```python
pool = pd.read_csv("Mandelbrot_pool.csv")
pool["time_s"] = pool["time_ns"] / 10**9

plt.plot(pool["pixels"], pool["time_s"], marker='o')
plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot with pool vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.show()
```

```python
many_pool = pd.read_csv("Mandelbrot_Many_Pools.csv")
many_pool["time_s"] = many_pool["time_ns"] / 10**9

plt.plot(many_pool["pixels"], many_pool["time_s"], marker='o')
plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot with pool vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.show()
```

```python
plt.plot(thread["pixels"], thread["time_s"], marker='o', label="Threads")
plt.plot(seq["pixels"], seq["time_s"], marker='o', markersize=5, label="Sequential")
plt.plot(pool["pixels"], pool["time_s"], marker='o', label="Pool")
plt.plot(many_pool["pixels"], many_pool["time_s"], marker='o', label="Many Pools")
plt.legend()
plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.show()
```

```python
max_sizes = [4, 8, 16, 32, 64, 128]
for max_size in max_sizes:
    many_pool = pd.read_csv(f"Mandelbrot_pool_size{max_size}.csv")
    many_pool["time_s"] = many_pool["time_ns"] / 10**9
    plt.plot(many_pool["pixels"], many_pool["time_s"], marker='o', label=f"max size {max_size}")

plt.xscale('log', base=2)
plt.yscale('log', base=2)
plt.title("Time to generate Mandelbrot with pool vs plot size")
plt.xlabel("Plot size (number of pixels)")
plt.ylabel("time (s)")
plt.legend()
plt.show()
```
